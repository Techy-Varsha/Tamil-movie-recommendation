# Tamil Movies Dataset Exploration

This notebook explores the Tamil movies dataset containing ~40 movies from 1992-2023.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('../data/tamil_movies.csv')
df.head()
```

```python
# Basic info
print(f"Dataset shape: {df.shape}")
print(f"Year range: {df['year'].min()} - {df['year'].max()}")
print(f"Total movies: {len(df)}")

df.info()
```

```python
# Lead actors distribution
actor_counts = df['lead_actor'].value_counts()
print(actor_counts)

plt.figure(figsize=(10, 6))
actor_counts.plot(kind='bar')
plt.title('Movies per Lead Actor')
plt.xlabel('Actor')
plt.ylabel('Number of Movies')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Genre distribution
genre_counts = df['genre'].value_counts()
print(genre_counts)

plt.figure(figsize=(8, 8))
genre_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title('Genre Distribution')
plt.ylabel('')
plt.show()
```

```python
# Box office trends over years
plt.figure(figsize=(12, 6))
sns.scatterplot(data=df, x='year', y='box_office_crores', hue='lead_actor', s=100)
plt.title('Box Office Collection by Year')
plt.xlabel('Year')
plt.ylabel('Box Office (Crores ₹)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Rating vs Box Office
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='rating', y='box_office_crores', hue='lead_actor', s=100)
plt.title('Rating vs Box Office Collection')
plt.xlabel('Rating (out of 10)')
plt.ylabel('Box Office (Crores ₹)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# Decade analysis
df['decade'] = (df['year'] // 10) * 10
decade_stats = df.groupby('decade').agg(
    count=('title', 'count'),
    avg_rating=('rating', 'mean'),
    total_box_office=('box_office_crores', 'sum'),
    avg_box_office=('box_office_crores', 'mean')
).round(2)
decade_stats
```

```python
# Top rated movies
top_rated = df.nlargest(10, 'rating')[['title', 'year', 'lead_actor', 'director', 'rating', 'box_office_crores']]
top_rated
```

```python
# Highest grossing movies
top_gross = df.nlargest(10, 'box_office_crores')[['title', 'year', 'lead_actor', 'director', 'rating', 'box_office_crores']]
top_gross
```

```python
# Director collaborations
director_actor = df.groupby(['director', 'lead_actor']).size().reset_index(name='count')
director_actor = director_actor.sort_values('count', ascending=False)
director_actor.head(15)
```

```python
# Music director analysis
music_stats = df.groupby('music_director').agg(
    movies=('title', 'count'),
    avg_rating=('rating', 'mean'),
    total_box_office=('box_office_crores', 'sum')
).round(2)
music_stats = music_stats.sort_values('movies', ascending=False)
music_stats.head(15)
```
